```python
import pandas as pd

# -----------------------------
# Load datasets
# -----------------------------
fund_master = pd.read_csv("01_fund_master.csv")
nav_history = pd.read_csv("02_nav_history.csv")

# -----------------------------
# Display first few rows
# -----------------------------
print("Fund Master Preview")
print(fund_master.head())

print("\nNAV History Preview")
print(nav_history.head())

# -----------------------------
# Display column names
# -----------------------------
print("\nFund Master Columns")
print(fund_master.columns.tolist())

print("\nNAV History Columns")
print(nav_history.columns.tolist())

# -----------------------------
# Automatically identify columns
# -----------------------------
fund_house_col = next((c for c in fund_master.columns if "house" in c.lower()), None)
category_col = next((c for c in fund_master.columns if "category" in c.lower()), None)
subcategory_col = next((c for c in fund_master.columns if "sub" in c.lower()), None)
risk_col = next((c for c in fund_master.columns if "risk" in c.lower()), None)
code_col = next((c for c in fund_master.columns if "code" in c.lower()), None)

nav_code_col = next((c for c in nav_history.columns if "code" in c.lower()), None)

print("\nDetected Columns")
print("Fund House :", fund_house_col)
print("Category :", category_col)
print("Sub Category :", subcategory_col)
print("Risk Grade :", risk_col)
print("AMFI Code :", code_col)

# -----------------------------
# Unique Values
# -----------------------------
if fund_house_col:
    print("\nUnique Fund Houses")
    print(fund_master[fund_house_col].unique())
    print("Total:", fund_master[fund_house_col].nunique())

if category_col:
    print("\nUnique Categories")
    print(fund_master[category_col].unique())
    print("Total:", fund_master[category_col].nunique())

if subcategory_col:
    print("\nUnique Sub Categories")
    print(fund_master[subcategory_col].unique())
    print("Total:", fund_master[subcategory_col].nunique())

if risk_col:
    print("\nUnique Risk Grades")
    print(fund_master[risk_col].unique())
    print("Total:", fund_master[risk_col].nunique())

# -----------------------------
# Validate AMFI Codes
# -----------------------------
if code_col and nav_code_col:

    master_codes = set(fund_master[code_col])
    nav_codes = set(nav_history[nav_code_col])

    missing = master_codes - nav_codes
    extra = nav_codes - master_codes

    print("\n========== DATA QUALITY ==========")
    print("Total Codes in Fund Master :", len(master_codes))
    print("Total Codes in NAV History :", len(nav_codes))
    print("Missing Codes :", len(missing))
    print("Extra Codes :", len(extra))

    if len(missing) > 0:
        print("\nMissing Codes")
        print(sorted(missing))

    if len(extra) > 0:
        print("\nExtra Codes")
        print(sorted(extra))

    print("\nData Quality Summary")

    if len(missing) == 0 and len(extra) == 0:
        print("✓ All AMFI codes are present in both datasets.")
        print("✓ No missing or extra scheme codes found.")
        print("✓ Data quality is good and suitable for analysis.")
    else:
        print("⚠ Dataset contains mismatched AMFI codes.")
        print("Investigate missing or extra codes before analysis.")
```
