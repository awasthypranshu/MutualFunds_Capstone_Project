```python
import requests
import pandas as pd

url = "https://api.mfapi.in/mf/125497"
```

```python
response = requests.get(url)
response.raise_for_status()
```

```python
result = response.json()
```

```python
meta = result["meta"]
df = pd.DataFrame(result["data"])


for key, value in meta.items():
    df[key] = value

df.to_csv("hdfc_top100_nav_with_metadata.csv", index=False)

print("CSV with metadata saved successfully!")
```
